```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import pandas as pd
```

```python
# Leer los datos desde un archivo CSV
data = pd.read_csv(
    r'D:\OneDrive\02_Carbon Market\03_Work\01_ALLCOT\07_Sembrando Futuro\Andean_BD.csv',sep=";")

print(data)
```

```python
# Suponiendo que las columnas son 'X' y 'Y'
X = data['NDVI_2020'].values.reshape(-1, 1)
y = data['AGB_2020'].values
```

```python
# Crear el modelo de regresión lineal
model = LinearRegression()

# Ajustar el modelo
model.fit(X, y)

# Obtener los parámetros del modelo
intercept = model.intercept_
slope = model.coef_[0]

# Predicciones
y_pred = model.predict(X)

# Calcular R²
r2 = r2_score(y, y_pred)

# Imprimir la ecuación de la recta y R²
print(f"Ecuación de la recta: y = {intercept:.2f} + {slope:.2f}x")
print(f"Coeficiente de determinación (R²): {r2:.2f}")

# Graficar los datos y la línea de regresión
plt.scatter(X, y, color='blue', label='Datos')
plt.plot(X, y_pred, color='red', label='Línea de regresión')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.show()
```
